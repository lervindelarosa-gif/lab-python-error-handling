   - If the user enters an invalid product name (e.g., a product name that is not in the inventory), or that doesn't have stock available, display an error message and ask them to re-enter the product name. *Hint: you will need to pass inventory as a parameter*

````python
def initialize_inventory(products):
    inventory = {}
    for product in products:
        valid_input = False
        while not valid_input:
            try:
                quantity = int(input(f"Enter the quantity of {product}s available: "))
                if quantity >= 0:
                    inventory[product] = quantity
                    valid_input = True
                else:
                    print("Quantity cannot be negative. Please enter a valid quantity.")
            except ValueError:
                print("Invalid input. Please enter a valid whole number.")
    return inventory


def get_customer_orders(inventory):
    # Get valid number of orders
    valid_num = False
    while not valid_num:
        try:
            num_orders = int(input("Enter the number of customer orders: "))
            if num_orders >= 0:
                valid_num = True
            else:
                print("Number of orders cannot be negative.")
        except ValueError:
            print("Invalid input! Please enter a whole number.")

    customer_orders = set()

    # Get valid product names
    for _ in range(num_orders):
        valid_product = False
        while not valid_product:
            product = input("Enter the name of a product that a customer wants to order: ").lower().strip()
            
            if product not in inventory:
                print(f"Error: '{product}' is not in the inventory. Please enter a valid product.")
            elif inventory[product] <= 0:
                print(f"Error: '{product}' is out of stock. Please select another product.")
            else:
                customer_orders.add(product)
                valid_product = True

    return customer_orders


def calculate_order_statistics(customer_orders, products):
    total_ordered = len(customer_orders)
    percentage_unique = (total_ordered / len(products)) * 100
    return total_ordered, percentage_unique


def print_order_statistics(order_statistics):
    total_ordered, percentage_unique = order_statistics
    print("\nOrder Statistics:")
    print(f"Total Products Ordered: {total_ordered}")
    print(f"Percentage of Unique Products Ordered: {percentage_unique:.1f}%")


def update_inventory(customer_orders, inventory):
    # Step 1: Subtract 1 for ordered items
    updated = {
        product: quantity - 1 if product in customer_orders else quantity
        for product, quantity in inventory.items()
    }
    # Step 2: Keep only items with stock remaining
    return {
        product: quantity 
        for product, quantity in updated.items() 
        if quantity > 0
    }


def print_updated_inventory(inventory):
    print("\nUpdated Inventory:")
    for product, quantity in inventory.items():
        print(f"{product}: {quantity}")


def calculate_total_price(customer_orders):
    total_price = 0.0
    for product in customer_orders:
        valid_price = False
        while not valid_price:
            try:
                price = float(input(f"Enter the price of {product}: "))
                if price >= 0:
                    total_price += price
                    valid_price = True
                else:
                    print("Price cannot be negative. Please enter a valid price.")
            except ValueError:
                print("Invalid input. Please enter a numeric value.")
    return total_price


# Main Program Execution
products = ["t-shirt", "mug", "hat", "book", "keychain"]

# 1. Initialize inventory with input validation
inventory = initialize_inventory(products)

# 2. Get customer orders with validation against inventory & stock
customer_orders = get_customer_orders(inventory)

# 3. Calculate and display order statistics
order_stats = calculate_order_statistics(customer_orders, products)
print_order_statistics(order_stats)

# 4. Update and display inventory
inventory = update_inventory(customer_orders, inventory)
print_updated_inventory(inventory)

# 5. Calculate and display total order price with input validation
total_price = calculate_total_price(customer_orders)
print(f"\nTotal Price: {total_price:.2f}")
````
